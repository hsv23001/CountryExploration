Beginning the exploratory data analysis with the combined dataset.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pylab as plt
```

```python
happy_index_15 = pd.read_csv("2015.csv")
happy_index_15.head()
```

```python
HIEF = pd.read_csv("HIEF_data.csv")
HIEF[HIEF.Year == 2013]
```

```python
HIEF_happy = HIEF[HIEF.Year == 2013].merge(happy_index_15, how = "left")
HIEF_happy.isna().sum()
```
